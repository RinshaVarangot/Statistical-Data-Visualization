## Matrix Plots

Matrix plots in Seaborn display data as color-encoded grids (matrices).
They help to:

* Visualize relationships between variables
* Identify patterns 🔍
* Detect clusters (similar groups)
* Understand correlations

The two main matrix plots are:

* Heatmap
* Clustermap

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
flights = sns.load_dataset('flights')
```

```python
tips=sns.load_dataset('tips')
```

```python
tips.head()
```

```python
flights.head()
```

## Heatmap

A heatmap shows matrix values using colors.

In order for a heatmap to work properly, your data should already be in a matrix form, the sns.heatmap function basically just colors it in for you. For example:

```python
tips.head()
```

```python
tips.corr(numeric_only=True)
```

```python
sns.heatmap(tips.corr(numeric_only=True))
plt.show()
```

```python
sns.heatmap(tips.corr(numeric_only=True),cmap='coolwarm',annot=True)
plt.show()
```

```python
flights.pivot_table(values='passengers',index='month',columns='year',observed=True)
```

```python
pvflights = flights.pivot_table(values='passengers',index='month',columns='year',observed=True)
sns.heatmap(pvflights)
plt.show()
```

```python
sns.heatmap(pvflights,cmap='magma',linecolor='white',linewidths=1)
plt.show()
```

## clustermap
The clustermap uses hierarchal clustering to produce a clustered version of the heatmap. For example:

```python
sns.clustermap(pvflights)
plt.show()
```

```python
# More options to get the information a little clearer like normalization
sns.clustermap(pvflights,cmap='coolwarm',standard_scale=1)
plt.show()
```

```python

sns.clustermap(pvflights,cmap='coolwarm',standard_scale=1,annot=True)
plt.show()
```
